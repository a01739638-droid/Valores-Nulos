```python
import pandas as pd
import numpy as np
```

```python
raw = pd.read_excel('SDC.xlsx', sheet_name='Hoja1', header=None)
raw.head(15)
```

DATASET 1: Indicadores Mensuales SDC

```python
header1 = raw.iloc[2, 0:37].tolist()
header1[35] = 'Total General'
header1[36] = 'Proporcion General'

data1 = raw.iloc[[3,4,5,6,7,9,10], 0:37].copy()
data1.columns = header1
data1 = data1.reset_index(drop=True)
data1
```

```python
valores_nulos = data1.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
data1['Total General'] = data1['Total General'].fillna(data1['Total General'].median())
```

```python
data1['Proporcion General'] = data1['Proporcion General'].fillna(round(data1['Proporcion General'].mean(), 4))
```

```python
valores_nulos = data1.isnull().sum().sum()
valores_nulos
```

```python
data1.to_csv('Indicadores_Mensuales_SDC.csv', index=False)
```

DATASET 2: Solicitudes por Financiera

```python
header2 = raw.iloc[2, 0:34].tolist()
header2[0] = 'Financiera'

data2 = raw.iloc[16:30, 0:34].copy()
data2.columns = header2
data2 = data2.reset_index(drop=True)
data2
```

```python
print('Cantidad de guiones (cero explicito):', (data2 == '-').sum().sum())
print('Cantidad de NaN reales:', data2.isnull().sum().sum())
```

```python
data2 = data2.replace('-', 0)
```

```python
columnas_mes = [c for c in data2.columns if c != 'Financiera']
for columna in columnas_mes:
    data2[columna] = data2[columna].fillna(0)
```

```python
valores_nulos = data2.isnull().sum().sum()
valores_nulos
```

```python
data2.to_csv('Solicitudes_por_Financiera_SDC.csv', index=False)
```
