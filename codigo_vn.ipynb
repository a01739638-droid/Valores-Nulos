```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Citas_Digital_Filtrado.csv')
data.head(5)
```

```python
data1= pd.read_csv('Citas_Digital_Filtrado.csv', skiprows=1)
data1
```

```python
valores_nulos = data.isnull().sum().sum()
valores_nulos
```

```python
data2 = data.drop(['Telefono '], axis=1)
data2.head(5)
```

```python
data3 = data2.copy()
```

```python
data3['BDC '] = data3['BDC '].ffill()
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['Fecha'] = data3['Fecha'].fillna('Sin fecha registrada')
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['Nombre Cliente '] = data3['Nombre Cliente '].fillna('Sin nombre registrado ')
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['Estatus de Lead'] = data3['Estatus de Lead'].bfill()
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['Potencial de compra'] = pd.to_numeric(data3['Potencial de compra'], errors='coerce')
data3['Potencial de compra'].isnull().sum()
```

```python
data3['Potencial de compra'] = data3['Potencial de compra'].fillna(round(data3['Potencial de compra'].mean(), 2))
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['PDM'] = data3['PDM'].fillna(data3['PDM'].median())
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['SDC'] = data3['SDC'].fillna(0)
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['Venta'] = data3['Venta'].fillna(data3['Venta'].mode()[0])
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['Asesor Asignado'] = data3['Asesor Asignado'].ffill()
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
columnas_motivo = ['¿Por que no ha visitado la agencia?', '¿Por que no hay solicitud de credito?',
                    '¿Por que no hubo prueba de manejo ?', '¿Por que no hubo proceso wow?',
                    '¿Por que se asigno antes de vistar piso?']

for columna in columnas_motivo:
    data3[columna] = data3[columna].fillna('No aplica')
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3['Unnamed: 15'] = data3['Unnamed: 15'].fillna('Sin comentarios adicionales')
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
valores_nulos = data3.isnull().sum().sum()
valores_nulos
```
